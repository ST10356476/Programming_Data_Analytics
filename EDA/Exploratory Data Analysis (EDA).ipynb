```python
# Identifying duplicate rows based on the 'Coaster Name' column
df.loc[df.duplicated(subset=['Coaster Name'])]
```

```python
# Checking an example of a specific roller coaster in the dataset using the query method. In this case, we are looking for the "Crystal Beach Cyclone" roller coaster.
df.query('`Coaster Name` == "Crystal Beach Cyclone"')
```

```python
df.columns
```

```python
# Removing duplicate rows based on the 'Coaster Name', 'Location', and 'Opening Date' columns, and resetting the index of the resulting DataFrame.
df = df.loc[~df.duplicated(subset=['Coaster Name', 'Location', 'Opening Date'])]     \
    .reset_index(drop=True)
```

# Step 3: Feature Understanding

(Univariate Analysis)

- Plotting Feature Distributions
    - Histogram

```python
# Counting the number of roller coasters introduced in each year
df['Year Introduced'].value_counts()
```

```python
# Counting the number of roller coasters introduced in each year
ax =df['Year Introduced'].value_counts() \
    .head(10) \
    .plot(kind='bar', title='Top Years with the Most Roller Coasters Introduced', xlabel='Year Introduced', ylabel='Number of Roller Coasters Introduced', figsize=(10, 6))
```

```python
ax =df['Speed_mph'].plot(kind='hist', bins=20, title='Distribution of Roller Coaster Speeds (mph)', xlabel='Speed (mph)', ylabel='Frequency', figsize=(10, 6))
```

```python
# Plotting the kernel density estimate of roller coaster speeds
ax = df['Speed_mph'].plot(kind='kde', title='Distribution of Roller Coaster Speeds (mph)', xlabel='Speed (mph)', ylabel='Density', figsize=(10, 6))
```

# Step 4: Feature Relationships

- Scatterplot
- Heatmap Correlation
- Pairplot
- Groupby Comparisons

```python
# Plotting a scatter plot of roller coaster speed vs height
df.plot(kind='scatter',
        x='Speed_mph',
        y='height_ft',
        title='Roller Coaster Speed vs Height',
        xlabel='Speed (mph)',
        ylabel='Height (ft)',
        figsize=(10, 6)
)
plt.show()
```

```python
# Plotting a scatter plot of roller coaster speed vs height with color coding by year introduced
sns.scatterplot(data=df, x='Speed_mph', y='height_ft', hue='Year Introduced')
plt.show()
```

```python
df.head()
```

```python
# Plotting a pair plot of selected variables with color coding by roller coaster type
sns.pairplot(df, vars=['Year Introduced', 'Speed_mph',
                      'height_ft', 'Inversions', 'Gforce'], hue='Type_Main')

plt.show()
```

```python
# Computing the correlation matrix for selected variables
df_corr = df[['Year Introduced', 'Speed_mph',
                      'height_ft', 'Inversions', 'Gforce']].dropna().corr()
```

```python
df_corr
```

```python
sns.heatmap(df_corr, annot=True)
```

# Step 5: Ask a Question about the data

- Try to answer a question you have about the data using a plot or statistic

For example what are the locations with the fastest roller coasters (Minimum of 10)?

```python
df.query('Location != "Other"') \
    .groupby('Location')['Speed_mph'] \
    .agg(['mean', 'count']) \
    .query('count >= 10') \
    .sort_values(by='mean')['mean'] \
    .plot(kind='barh', title='Average Roller Coaster Speed by Location (for Locations with at least 10 Roller Coasters)', xlabel='Location', ylabel='Average Speed (mph)', figsize=(12, 6))
```
